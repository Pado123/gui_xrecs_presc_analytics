```python
import os
import pandas as pd

experiments_dir = "experiments"

def parse_csv_log(file_path):
    activities = set()
    cases = set()
    try:
        df = pd.read_csv(file_path)
        # The activity column is usually 'concept:name'
        if 'concept:name' in df.columns:
            activities.update(df['concept:name'].dropna().unique())
        # Case column: most commonly 'case:concept:name' or 'case_id'
        if 'case:concept:name' in df.columns:
            cases.update(df['case:concept:name'].dropna().unique())
        elif 'case_id' in df.columns:
            cases.update(df['case_id'].dropna().unique())
    except Exception as e:
        print(f"Error parsing {file_path}: {e}")
    return activities, cases

log_stats = {}

for root_dir, _, files in os.walk(experiments_dir):
    for file in files:
        if file.endswith('.csv'):
            file_path = os.path.join(root_dir, file)
            activities, cases = parse_csv_log(file_path)
            log_stats[file_path] = {
                'num_activities': len(activities),
                'num_cases': len(cases)
            }

for log, stats in log_stats.items():
    print(f"Log: {log}")
    print(f"  Number of different activities: {stats['num_activities']}")
    print(f"  Number of different cases: {stats['num_cases']}")
    print()
```
